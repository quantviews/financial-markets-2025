# Типы данных

значениеs have Типы данных. You can think of a type as a group of similar значениеs. Example: all interger значениеs are members of type 'int'.

These Типы данных are provided out of the box:

- `int`
- `число с плавающей точкой`
- `str`
- `bool`
- `None`

Python allows you not to think of Типы данных at all - and it is good for start.

- hard to reason about Типы данных before you explore your task well
- not always you need 5-7 Типы данных of `целое число`
- type notation can grow very cumbersome

But why bother about Типы данных?

- better structure and stability of the programs
- more guarantees, easier to build large programs
- understanding error messages
- better computer representation of data
- programs usually runs faster
- may have different операции on different Типы данных (eg "multiple dispatch" in Julia)
- strong [theory about Типы данных](https://thinkingwithtypes.com/)

Python is a dynamically typed language, but there are _type annotations_ that improve readibilty of the program and can be checked by external tools (eg mypy).

Terminology:
- static typing (Rust, nimlang, vlang, Haskell)
- dynamic typing (Python, Ruby, Julia)
